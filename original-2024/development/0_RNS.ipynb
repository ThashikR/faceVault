```python
import numpy as np
from PIL import Image

# Function to encrypt an image using RNS
def encrypt_image(image_path, moduli):
    # Load image
    img = Image.open(image_path).convert('L')
    img_array = np.array(img)
    
    # Encrypt each pixel value
    encrypted_img_array = np.zeros(img_array.shape + (len(moduli),), dtype=np.uint8)
    for i in range(img_array.shape[0]):
        for j in range(img_array.shape[1]):
            pixel_value = img_array[i, j]
            encrypted_pixel = [pixel_value % modulus for modulus in moduli]
            encrypted_img_array[i, j] = encrypted_pixel
    
    return encrypted_img_array
    

# Function to decrypt an image using RNS
def decrypt_image(encrypted_img_array, moduli):
    # Decrypt each pixel value
    decrypted_img_array = np.zeros_like(encrypted_img_array)
    for i in range(encrypted_img_array.shape[0]):
        for j in range(encrypted_img_array.shape[1]):
            encrypted_pixel = encrypted_img_array[i, j]
            decrypted_pixel = sum([encrypted_pixel[k] * moduli[k] for k in range(len(moduli))]) % np.prod(moduli)
            decrypted_img_array[i, j] = decrypted_pixel
    
    # Convert decrypted pixel values to uint8
    decrypted_img_array = decrypted_img_array.astype(np.uint8)
    
    return decrypted_img_array

# Example usage

# Define moduli for the RNS system
moduli = [251, 257, 263]  # Example moduli
image_path = "new.jpg"

# Encrypt image
encrypted_img_array = encrypt_image(image_path, moduli)
encrypted_img = Image.fromarray(encrypted_img_array)
encrypted_img.show()

# Decrypt image
decrypted_img_array = decrypt_image(encrypted_img_array, moduli)

# Convert decrypted array back to image and display
decrypted_img = Image.fromarray(decrypted_img_array)
decrypted_img.show()

```
